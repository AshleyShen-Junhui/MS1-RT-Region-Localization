```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# =========================
# 1. load data
# =========================
# csv_path = Path("results/final/all_ckpt_metrics_STP1710.7_vs_control.csv").expanduser()


PROJECT_ROOT = Path("../..").resolve()
csv_path=PROJECT_ROOT /"results"/"final"/"all_ckpt_metrics__STP1710.7_vs_control__c_sweep_log19.csv"
save_dir=PROJECT_ROOT /"figures"/"qc"




df = pd.read_csv(csv_path)

df = df[df["step"].notna()].copy()
df["step"] = df["step"].astype(int)

metric_col = "AUROC_STP1710.7_vs_control"

# 如有多个配置（如C），请固定一个
# df = df[df["C"] == 40].copy()

# =========================
# 2. aggregate
# =========================
summary = (
    df.groupby("step")[metric_col]
      .agg(["mean", "std", "count"])
      .reset_index()
      .sort_values("step")
)

summary["se"] = summary["std"] / np.sqrt(summary["count"])
summary["ci95"] = 1.96 * summary["se"]
summary["lower"] = summary["mean"] - summary["ci95"]
summary["upper"] = summary["mean"] + summary["ci95"]

# =========================
# 3. style（最终版参数）
# =========================
MAIN_COLOR = "#2A5D9F"     # 主线（更深蓝）
FILL_COLOR = "#6FA8DC"     # CI（浅蓝）
BASELINE_COLOR = "#B5B5B5" # baseline（更淡灰）

plt.figure(figsize=(7.2, 4.8))  # 双栏友好尺寸

# ---- CI ----
plt.fill_between(
    summary["step"],
    summary["lower"],
    summary["upper"],
    color=FILL_COLOR,
    alpha=0.20,     # ⚠️关键：更轻
    linewidth=0,
    zorder=1
)

# ---- mean line ----
plt.plot(
    summary["step"],
    summary["mean"],
    color=MAIN_COLOR,
    marker="o",
    markersize=5.5,
    linewidth=2.8,
    markerfacecolor=MAIN_COLOR,
    markeredgecolor="white",
    markeredgewidth=1.2,
    zorder=3
)

# ---- baseline ----
plt.axhline(
    0.5,
    linestyle="--",
    linewidth=1.0,
    color=BASELINE_COLOR,
    zorder=2
)

# ---- grid（很轻）----
plt.grid(
    axis="y",
    linestyle="--",
    linewidth=0.6,
    alpha=0.18
)

# =========================
# 4. labels
# =========================
plt.title(
    "STP1710.7 vs control",
    fontsize=14,
    pad=10
)
plt.xlabel("Checkpoint step", fontsize=14)
plt.ylabel("AUROC", fontsize=14)

plt.xticks(summary["step"], fontsize=12)
plt.yticks(fontsize=12)

# ⚠️收紧范围（更专业）
#plt.ylim(0.3, 0.80)


# 去掉上右边框（论文风）
ax = plt.gca()
ax.set_xlim(50, 700)
ax.set_xticks([100,200,300,400,500,600,700])
ax.set_ylim(0.3, 0.75)
ax.set_yticks([0.4, 0.5, 0.6, 0.7])
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_linewidth(1.0)
ax.spines["bottom"].set_linewidth(1.0)

plt.tight_layout()

# =========================
# 5. save
# =========================
#plt.savefig("Figures/checkpoint_dynamics_STP1710_final.png", dpi=300, bbox_inches="tight")
save_path = save_dir / "checkpoint_dynamics_STP1710_final.png"
plt.savefig(save_path, dpi=300, bbox_inches="tight")
plt.show()
```
